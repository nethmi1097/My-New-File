```python
import pandas as pd

data = pd.read_csv("Student_performance_data _.csv")
data.head()
```

```python
import matplotlib.pyplot as plt


plt.plot(data["GPA"].head(10), label="GPA")
plt.plot(data["StudyTimeWeekly"].head(10),label="Study Time Weekly")

plt.title("Student Score Comparison")
plt.xlabel("Students")
plt.ylabel("Scores")

plt.legend()
plt.show()
```

```python
gender_counts = data["Gender"].value_counts()
gender_counts
```

```python
gender_counts = data["Gender"].value_counts()

gender_counts.plot(kind="bar")

plt.title("Number of students by gender")
plt.xlabel("Gender")
plt.ylabel("Count")

plt.show()
```

```python

avg_gpa = data.groupby("Gender")["GPA"].mean()

avg_gpa.plot(kind="bar")

plt.title("Average GPA by Gender")
plt.xlabel("Gender")
plt.ylabel("Average GPA")

plt.show()
```

```python
import matplotlib.pyplot as plt

plt.scatter(data["StudyTimeWeekly"],data["GPA"])

plt.title("Study Time vs GPA")
plt.xlabel("Weekly Study Time")
plt.ylabel("GPA")

plt.show()
```

```python
corr_matrix = data.corr()

print(corr_matrix)
```

```python
import seaborn as sns

plt.figure(figsize=(20,10))

sns.heatmap(data.corr() , annot=True)

plt.title("Correlation Heatmap")

plt.show()
```
